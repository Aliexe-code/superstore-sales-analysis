# Analysis Questions

## Question 1
**Question:** Which product categories and regions generate the most revenue and profit?
**Why it matters:** Identifying top-performing categories and regions helps
the business decide where to focus inventory, marketing, and sales effort.
It also reveals whether any category or region drives high sales but low profit.
**Chart I expect to use:** Bar chart (sales/profit by category and by region).

## Question 2
**Question:** How have sales and profit changed over time (monthly/yearly)?
**Why it matters:** Understanding the trend over the 4-year period shows
whether the business is growing, shrinking, or seasonal, which affects
forecasting and planning decisions.
**Chart I expect to use:** Line chart (sales/profit over time).

## Question 3
**Question:** How does discount level relate to profit?
**Why it matters:** Discounts drive sales volume but can erode margins.
Understanding this relationship helps the business set discount policy
without unintentionally destroying profit.
**Chart I expect to use:** Scatter plot (discount % vs. profit), possibly
supported by a correlation coefficient.
